## __Using Ollama Library__

```python
import ollama
```

**Generate Responses**

```python
## Generation
response = ollama.generate(model="llama3.2:latest", prompt="Why does moon glow")

print(response)
```

```python
print(response.response)
```

```python
## stream response

response = ollama.generate(model="llama3.2:latest", prompt="Why does moon glow", stream=True)

for i in response:
    print(i["response"], end="")
```

```python
## Giving multi model input to model (Passing to the Local LLM Model)

import base64
import ollama

image_path = "Linkedin.jpg"

with open(image_path, "rb") as f:  # Opening the image in read binary
    image_bytes = f.read()

image_64 = base64.b64encode(image_bytes).decode("utf-8")  ## Encoding the image in base64 bits

response = ollama.generate(model="gemma3:4b", images=[image_64], prompt="Give caption for the image")
print(response.response)
```

```python
# PASSING MULTIPLE IMAGES TO THE MODEL.

image_paths = ["Linkedin.jpg","Green AI.png"]

images_base64=[]
for i in image_paths:
    with open(i , "rb") as f:
        image_bytes= f.read()
        images_base64.append(base64.b64encode(image_bytes).decode("utf-8"))

response= ollama.generate(model="gemma3:4b", images=images_base64,
                          prompt="Generate an story based on these images, make sure you take context from each and every image.")
print(response.response)
```

```python
response = ollama.generate(model="llama3.2:latest", prompt="Why does moon glow?", 
                           system="You are a funny assistant, you explain things in a funny way")

print(response.response)
```

```python
## Options Parameter

response = ollama.generate(
    model="llama3.2:latest", 
    prompt="Why is the ocean blue ?",
    options={
        "temperature": 0.3,
        "top_p": 0.5,
        "top_k": 45
    }
)

print(response.response)
```
