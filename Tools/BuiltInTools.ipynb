## A session about **BuiltIn Tools**

__1. BuiltIn tool - DuckDuckGo__

```python
from ddgs import DDGS
from langchain_core.tools import tool

@tool
def duckduckgo_search(query: str) -> str:
    """Search the internet using DuckDuckGo."""

    with DDGS() as ddgs:
        results = list(
            ddgs.text(query, max_results=5)
        )

    return "\n".join(
        f"Title: {r['title']}\n"
        f"URL: {r['href']}\n"
        f"Content: {r['body']}"
        for r in results
    )

result = duckduckgo_search.invoke({"query": "IPL News"})

print(result)
```

__2. Builtin Tools - Shell Tool__

```python
from langchain_community.tools import ShellTool

shell_tool = ShellTool()
results = shell_tool.invoke("whoami")
res = shell_tool.invoke("dir")

print(results)
print(res)
```
