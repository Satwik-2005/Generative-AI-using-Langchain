## Generation of **Custom Tools**

```python
from langchain_core.tools import tool
```

```python
## step 1: Create a function

def multiply(a, b):
    """Multiply two Numbers"""
    return (a * b)
```

```python
## step 2: Type Hinting

def multiply(a: int, b: int) -> int:
    """Multiply two Numbers"""
    return (a * b)    
```

```python
## step 3: Add @tool decorator

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two Numbers"""
    return (a * b)    
```

```python
## using of tool

result = multiply.invoke({"a": 3, "b": 5})
```
