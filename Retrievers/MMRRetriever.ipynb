## MMR Retriever

```python
from langchain_core.documents import Document
```

```python
# Sample documents
docs = [
    Document(page_content="LangChain makes it easy to work with LLMs."),
    Document(page_content="LangChain is used to build LLM based applications."),
    Document(page_content="Chroma is used to store and search document embeddings."),
    Document(page_content="Embeddings are vector representations of text."),
    Document(page_content="MMR helps you get diverse results when doing similarity search."),
    Document(page_content="LangChain supports Chroma, FAISS, Pinecone, and more."),
]
```

```python
from langchain_community.vectorstores import FAISS
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
## Initialize the Embedding Models
embeddings = HuggingFaceEmbeddings(model_name='sentence-transformers/all-MiniLM-L6-v2')

## Create the FAISS vectorstore from documents
vector_store = FAISS.from_documents(
    documents=docs,
    embedding=embeddings
)
```

```python
## Enable the MMR retriever
retriever = vector_store.as_retriever(
    search_type="mmr",   ## <--- This enables to use MMR retriever
    search_kwargs={"k": 3, "lambda_mult": 0.5}   ### <---  k = top_results, lambda_mult = diverse results (different results)  1 -> same as similarity search, 0 -> pure different results
)
```

```python
query = "What is langchain?"

results = retriever.invoke(query)
```

```python
for i, doc in enumerate(results):
    print(f"\n ---  Result: {i + 1}  ---")
    print(doc.page_content)
```
